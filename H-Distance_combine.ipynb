```python
import matplotlib.pyplot as plt
import numpy as np

# Function to read XVG file and extract data
def read_xvg(file_path):
    data = np.genfromtxt(file_path, skip_header=17)  # Adjust skip_header based on your XVG file format
    time_points = data[:, 0]
    rmsd_values = data[:, 1]
    return time_points, rmsd_values

# Replace these placeholders with your actual file paths
D1_path = '47-UNK-O11-180LEU-H-distance.xvg'
D2_path = '48-UNK-H5-177ALA-O-distance.xvg'
D3_path = '50-UNK-O8-182GLY-H-distance.xvg'
D4_path = '53-UNK-O8-181THR-H-distance.xvg'

# Read data from XVG files
time_D1, D1_data = read_xvg(D1_path)
time_D2, D2_data = read_xvg(D2_path)
time_D3, D3_data = read_xvg(D3_path)
time_D4, D4_data = read_xvg(D4_path)

# Create a new figure
fig, ax = plt.subplots()

# Plot the data
ax.plot(time_D1, D1_data, label='UNK(O11) - 180LEU(H)', color='blue')
ax.plot(time_D2, D2_data, label='UNK(H5) - 177ALA(O)', color='red')
ax.plot(time_D3, D3_data, label='UNK(O8) - 182GLY(H)', color='forestgreen')
ax.plot(time_D4, D4_data, label='UNK(O8) - 181THR(H)', color='deepskyblue')

# Set axis limits
ax.set_xlim(0, 100)
ax.set_ylim(0, 1)

# Add labels and legend with custom font style
ax.set_xlabel('Time (ns)', fontsize=16)  # Change 'Arial' to your desired font
ax.set_ylabel('H-Bond Distance (nm)', fontsize=16)  # Change 'Arial' to your desired font
ax.legend(fontsize=10)

# Adjust number of markings on the x and y axes
ax.xaxis.set_major_locator(plt.MaxNLocator(4))  # Adjust the number of markings on x-axis
ax.yaxis.set_major_locator(plt.MaxNLocator(10))  # Adjust the number of markings on y-axis

# Adjust the marking size
ax.tick_params(axis='both', which='major', labelsize=14)  # Adjust the size of markings

# Save the plot as an SVG file
plt.savefig('H-bond_distance.svg', format='svg', bbox_inches='tight')

# Show the plot
plt.show()
```
